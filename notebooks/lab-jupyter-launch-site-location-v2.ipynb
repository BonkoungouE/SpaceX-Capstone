# Task 2: Mark the success/failed launches for each site on the map


Next, let's try to enhance the map by adding the launch outcomes for each site, and see which sites have high success rates.
Recall that data frame spacex_df has detailed launch records, and the `class` column indicates if this launch was successful or not


```python
spacex_df.tail(10)
```

Next, let's create markers for all launch records. 
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

_TODO:_ Create a new column in `launch_sites` dataframe called `marker_color` to store the marker colors based on the `class` value


```python
spacex_df['marker_color'] = spacex_df['class'].map({1: 'green', 0: 'red'})
```

```python
# Function to assign color to launch outcome
def assign_marker_color(launch_outcome):
    if launch_outcome == 1:
        return 'green'
    else:
        return 'red'
    
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)
spacex_df.tail(10)
```

_TODO:_ For each launch result in `spacex_df` data frame, add a `folium.Marker` to `marker_cluster`


```python
marker_cluster = MarkerCluster(name='Launch outcomes').add_to(site_map)
for index, record in spacex_df.iterrows():
    outcome = 'Success' if record['class'] == 1 else 'Failure'
    folium.Marker([record['Lat'], record['Long']],
        popup=f"Launch {index + 1} — {escape(record['Launch Site'])} — {outcome}",
        icon=folium.Icon(color='white', icon_color=record['marker_color'], icon='info-sign')).add_to(marker_cluster)
site_map
```

# TASK 3: Calculate the distances between a launch site to its proximities


Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.


You can calculate the distance between two points on the map based on their `Lat` and `Long` values using the following method:


```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

_TODO:_ Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# Reference point supplied in the exercise, not an exact shortest-distance estimate.
reference_site = launch_sites_df.loc[launch_sites_df['Launch Site'].eq('CCAFS SLC-40')].iloc[0]
launch_site_lat, launch_site_lon = float(reference_site['Lat']), float(reference_site['Long'])
coastline_lat, coastline_lon = 28.56367, -80.57163
coastline_coordinate = [coastline_lat, coastline_lon]
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f'Distance to selected coastline reference: {distance_coastline:.2f} km')
```

_TODO:_ After obtained its coordinate, create a `folium.Marker` to show the distance


```python
distance_marker = folium.Marker(coastline_coordinate,
    popup='Coastline reference supplied in lab (approximate)',
    icon=DivIcon(icon_size=(220, 30), html=f'<div style="font-size:12px;color:#d35400"><b>Coast reference: {distance_coastline:.2f} km</b></div>'))
site_map.add_child(distance_marker)
```

_TODO:_ Draw a `PolyLine` between a launch site to the selected coastline point


```python
lines = folium.PolyLine([[launch_site_lat, launch_site_lon], coastline_coordinate],
                        weight=2, color='#d35400', tooltip=f'{distance_coastline:.2f} km')
site_map.add_child(lines)
```

```python
# Approximate study coordinates: verify their feature assignment using MousePosition.
# These points are examples, not a search for the nearest road/railway/city boundary.
reference_points = {
    'Railway study point (verify)': [28.5721, -80.5852],
    'Road study point (verify)': [28.5630, -80.5768],
    'City study point (verify)': [28.3922, -80.6077],
}
proximity_records = []
for name, coordinate in reference_points.items():
    distance = calculate_distance(launch_site_lat, launch_site_lon, *coordinate)
    proximity_records.append({'Reference point': name, 'Latitude': coordinate[0],
                              'Longitude': coordinate[1], 'Distance_km': distance})
    folium.Marker(coordinate, popup=f'{name}: {distance:.2f} km',
                  icon=folium.Icon(color='blue')).add_to(site_map)
    folium.PolyLine([[launch_site_lat, launch_site_lon], coordinate], color='blue',
                    weight=2, tooltip=f'{name}: {distance:.2f} km').add_to(site_map)
pd.DataFrame(proximity_records)
```

### Observations et limites
Les sites du jeu de données sont proches des côtes, en Floride et en Californie. Ils ne se trouvent pas sur l’équateur. Une ouverture vers l’océan permet des trajectoires au-dessus de zones moins habitées. La carte est descriptive et n’établit pas de causalité avec la réussite.

Les distances sont calculées à vol d’oiseau (formule de Haversine). Le point côtier provient de l’exemple fourni ; les autres coordonnées sont des points d’étude approximatifs dont l’affectation doit être vérifiée sur la carte avec MousePosition. Elles ne représentent pas des distances minimales validées à la côte, au réseau routier, au chemin de fer ou à une limite urbaine. Les limites des villes et l’accès réel aux infrastructures exigent des données géographiques supplémentaires.

Le fond cartographique et les bibliothèques JavaScript de la carte nécessitent une connexion Internet. Le fichier HTML exporté peut être ouvert dans un navigateur.
